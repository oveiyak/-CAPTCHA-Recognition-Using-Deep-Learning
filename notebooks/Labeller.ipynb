```python
def copy_directory_structure(src, dst):
    for root, dirs, files in os.walk(src):
        # Construct destination path by replacing the source root with destination root
        # os.path.relpath(root, src) gets all the paths relative to src
        # print(root)
        # print(dirs)
        # print(files)
        dest_dir = os.path.join(dst, os.path.relpath(root, src))
        print(f'makedir: {dest_dir}')
        os.makedirs(dest_dir, exist_ok=True)  # Create each directory if it doesn't exist
```

```python
import os
import shutil
from PIL import Image
import matplotlib.pyplot as plt
import time

# Function to load images from the relevant folder and allow the user to label them
```

```python
def label_captchas(user_name, captcha_type, shared_folder):
    folder_path = f"{base_path}captchas/{user_name}"  # Folder based on user's first name
    labels_file = f"{folder_path}/labels.csv"  # File to store labels
    # Check if folder exists
    if not os.path.exists(folder_path):
        print(f"No folder found for {user_name}. Please check the folder name.")
        return
    # Load list of images
    images = [f for f in os.listdir(folder_path) if f.endswith('.png')]
    # Check if there are images to label
    if len(images) == 0:
        print(f"No CAPTCHA images found in {folder_path}.")
        return
    # Check if labels file already exists to continue from last labeled image
    labeled_images = []
    if os.path.exists(labels_file):
        with open(labels_file, 'r') as f:
            labeled_images = [line.split(',')[0] for line in f.readlines()]
    # Open file in append mode to save labels
    with open(labels_file, 'a') as f:
        for image_file in images:
            # Skip images that are already labeled
            if image_file in labeled_images:
                continue
            # Display the CAPTCHA image using PIL and matplotlib
            img_path = os.path.join(folder_path, image_file)
            img = Image.open(img_path)
            display(img)
            time.sleep(0.1)
            # Ask the user for the CAPTCHA label
            label = input(f"Enter label for {image_file}: ")
            label = label.upper() if captcha_type else label
            if label == '0':
                print("Haulted with 0 entry")
                return
            # Save the label in the format: image_file,label
            f.write(f"{image_file},{label}\n")
            print(f"Label for {image_file} saved.")
            # Rename the image file with the label and move it to the shared folder
            new_image_name = f"{label}.png"
            new_image_path = os.path.join(f'{shared_folder}{user_name}', new_image_name)
            shutil.move(img_path, new_image_path)  # Move the image to the shared folder
            print(f"Image {image_file} renamed to {new_image_name} and moved to {shared_folder}{user_name}.")
            # Close the image display
            plt.close()
    print("All CAPTCHA images have been labeled and moved to the shared folder.")

```

```python

def main():
    # user_name = input("Enter your first name: ").lower()
    captcha_type = int(input("Are all captchas in CAPS? Enter 1 for yes, else 0: "))
    # shared_folder = f'{base_path}labeled_captchas/'  # Path to the shared folder
    print("\nNOW YOU WILL ENTER A LOOP OF CAPTCHAS YOU HAVE TO LABEL.")
    print("TO EXIT ENTER 0 IN PLACE OF THE CAPTCHA VALUE")
    print("Data will be constanttly stored and updated to a csv file as well as the images will be renamed and stored in your folders.")
    label_captchas(user_name, captcha_type, label_captcha_path)

if __name__ == "__main__":
    main()
```
